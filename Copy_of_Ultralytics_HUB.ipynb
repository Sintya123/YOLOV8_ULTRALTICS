```python
# 1. Install Ultralytics jika belum
!pip install ultralytics

# 2. Import library yang dibutuhkan
from ultralytics import YOLO
from google.colab import files
import os
from IPython.display import Image, display
import glob

# 3. Load model yang sudah kamu latih
model = YOLO('runs/detect/train/weights/best.pt')  # ganti path jika berbeda

# 4. Upload gambar untuk dideteksi
uploaded = files.upload()

# 5. Simpan gambar ke folder predict_input/
os.makedirs('predict_input', exist_ok=True)
for filename in uploaded.keys():
    os.rename(filename, f'predict_input/{filename}')

# 6. Jalankan deteksi menggunakan YOLOv8
results = model.predict(source='predict_input', save=True, save_txt=True, project='runs', name='detect_result')

# 7. Ambil dan tampilkan hasil deteksi
latest_result_dir = 'runs/detect/detect_result'
image_paths = glob.glob(f'{latest_result_dir}/*.jpg')

for img_path in image_paths:
    display(Image(filename=img_path, width=700))
```

```python
import cv2
import os
import matplotlib.pyplot as plt

# Path ke folder hasil deteksi
result_path = 'runs/detect_result'

# Membaca gambar di dalam folder tersebut
for filename in os.listdir(result_path):
    file_path = os.path.join(result_path, filename)

    if file_path.endswith(('.jpg', '.png', '.jpeg')):  # Pastikan hanya gambar yang dibaca
        # Baca gambar menggunakan OpenCV
        img = cv2.imread(file_path)

        # Mengubah gambar dari BGR ke RGB untuk ditampilkan dengan matplotlib
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Tampilkan gambar
        plt.imshow(img_rgb)
        plt.axis('off')  # Menyembunyikan axis
        plt.show()
```
